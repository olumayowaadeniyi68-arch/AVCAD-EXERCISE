```python
import matplotlib.pyplot as plt

fruits = ["Apples", "Bananas", "Oranges", "Grapes"]
sales = [50, 80, 40, 70]

plt.bar(fruits, sales)

plt.title("Fruit Sales")
plt.xlabel("Fruit Type")
plt.ylabel("Number of Sales")

plt.show()
```
