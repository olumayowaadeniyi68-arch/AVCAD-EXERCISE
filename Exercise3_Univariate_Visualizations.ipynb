```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("EFIplus_medit.csv")

# Show
```

```python
# Show column names
print(df.columns)
```

```python
# Find top 4 catchments
top_catchments = df['Catchment_name'].value_counts().head(4)

# Show results
print(top_catchments)
```

```python
# Store top 4 catchment names
top4 = top_catchments.index

print(top4)
```

```python
# Create filtered dataset using top 4 catchments
filtered_df = df[df['Catchment_name'].isin(top4)]

# Preview filtered dataset
filtered_df.head()
```
